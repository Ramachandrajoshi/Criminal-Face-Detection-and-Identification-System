# Model Comparison

Compare face recognition models available through `deepface`:

| Model | Embedding Dims | Loss | Use-case |
|---|---|---|---|
| **ArcFace** (default) | 512 | Additive Angular Margin | Primary: occlusion-robust |
| FaceNet | 128 / 512 | Triplet Loss | Benchmark |
| VGG-Face | 4096 | Softmax / Cross-Entropy | Benchmark: high-quality frontal |
| SFace | 512 | Sigmoid-Constrained Hypersphere | Benchmark: low-res CCTV |

## Evaluation Metrics

1. **Embedding dimension** — lower is better for storage and query speed
2. **Cosine similarity accuracy** — measured on LFW/CASIA-WebFace
3. **Inference time** — time to extract embedding from aligned face
4. **Robustness to occlusion** — performance with mask, sunglasses, partial face

## Default: ArcFace

Per AGENTS.md, ArcFace is the default model:
- 512-d embeddings (compact storage)
- Additive Angular Margin loss (superior discriminative power)
- Best performance in the presence of partial occlusions
