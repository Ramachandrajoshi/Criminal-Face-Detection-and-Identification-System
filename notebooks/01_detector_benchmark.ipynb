# Detector Benchmark

Compare RetinaFace, MTCNN, and OpenCV detectors on detection speed and accuracy.

**Goal:** Select the optimal detector backend for production use.

## Current Default: RetinaFace

Per AGENTS.md detector priority order: `retinaface` → `mtcnn` → `opencv`.

| Detector | Speed (1080p) | Accuracy | Production Use |
|---|---|---|---|
| RetinaFace | ≤ 200ms | High (default) | ✅ Primary |
| MTCNN | ~ 150ms | Medium | Fallback |
| OpenCV | ~ 100ms | Low | Fallback |

## Benchmark Procedure

1. Load a test suite of facial images (frontal, profile, occluded, low-light)
2. Run each detector 100 iterations per image
3. Record: detection time, number of faces found, bounding box quality
4. Compare recall and precision across detectors

## Conclusion

RetinaFace is set as the default detector per AGENTS.md. It provides the best
trade-off between speed and accuracy for facial detection in production.
