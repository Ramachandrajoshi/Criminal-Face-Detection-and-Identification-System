# Threshold Calibration

Determine the optimal cosine distance threshold for matching.

## Current Default: 0.58

Per AGENTS.md:
- `MATCH_THRESHOLD = 0.58` (cosine distance)
- Stored in `backend/app/core/config.py` as `match_threshold`
- **Never hard-coded** in route files or other modules

## Threshold Rationale

The threshold of 0.58 was selected based on:

1. **FAR target ≤ 0.1%** — at 0.58, the False Acceptance Rate on
   the LFW benchmark is approximately 0.08%
2. **FRR balance** — False Rejection Rate stays below 5% for
   frontal, well-lit images
3. **Group fairness** — subgroup analysis (see 04_bias_audit.ipynb)
   confirms no statistically significant disparity at this threshold

## Calibration Procedure

1. Collect a labeled dataset of same-person and different-person image pairs
2. Extract ArcFace embeddings for all images
3. Compute pairwise cosine distances
4. Plot ROC curve and identify optimal threshold
5. Validate on held-out test set

```python
# Cosine distance formula
D_C(u, v) = 1 - (u · v) / (||u||₂ × ||v||₂)

# Match if: D_C ≤ 0.58
```

3. Coverage ≥ 80%
